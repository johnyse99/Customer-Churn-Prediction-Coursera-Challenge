```python
# ==============================================================================
# PROJECT: Customer Churn Prediction (Strict Lab Requirements)
# SCRIPT: Single-Cell Style for Jupyter Notebook
# ==============================================================================

import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier  # Reemplaza Dummy
from sklearn.preprocessing import LabelEncoder
from matplotlib import pyplot as plt
import seaborn as sns

# Visualization Packages
%matplotlib inline
```

```python
# --- LOAD THE DATA ---
# Import required packages
data_descriptions = pd.read_csv("../data/data_descriptions.csv")
pd.set_option("display.max_colwidth", None)
print("Data Descriptions:")
display(data_descriptions)
```

```python
train_df = pd.read_csv("../data/train.csv")
print("train_df Shape:", train_df.shape)
```

```python
test_df = pd.read_csv("../data/test.csv")
print("test_df Shape:", test_df.shape)
```

```python
# --- EXPLORE, CLEAN AND VALIDATE (Optional) ---
# Essential preprocessing for the model to function within the lab
def preprocess_for_lab(df):
    temp_df = df.copy()
    le = LabelEncoder()
    # Codificamos solo columnas categóricas que no sean IDs
    for col in temp_df.select_dtypes(include=["object"]).columns:
        if col != "CustomerID":
            temp_df[col] = le.fit_transform(temp_df[col].astype(str))
    return temp_df


train_clean = preprocess_for_lab(train_df)
test_clean = preprocess_for_lab(test_df)
```

```python
# --- MAKE PREDICTIONS (REQUIRED) ---
# Requirement: dataframe 'prediction_df' with 104,480 entries
# Requirement: Exact columns 'CustomerID' and 'predicted_probability'

### PLEASE CHANGE THIS CODE TO IMPLEMENT YOUR OWN PREDICTIONS

# 1. Define features and target based on train_df
X = train_clean.drop(["CustomerID", "Churn"], axis=1)
y = train_clean.Churn
```

```python
# 2. Train the model (Random Forest to ensure good AUC)
model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
model.fit(X, y)
```

```python
# 3. Making predictions in test_df using predict_proba
# Extraemos la probabilidad de la clase 1 (Churn)
predicted_probability = model.predict_proba(test_clean.drop(["CustomerID"], axis=1))[
    :, 1
]
```

```python
# 4. Combine predictions in the dataframe with exact names
prediction_df = pd.DataFrame(
    {
        "CustomerID": test_df[["CustomerID"]].values[:, 0],
        "predicted_probability": predicted_probability,
    }
)
```

```python
# --- VIEW RESULTS ---
# It must contain 104,480 rows and 2 columns
print("Shape final:", prediction_df.shape)
display(prediction_df.head(10))
```
